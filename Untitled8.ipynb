```python
from IPython.display import Audio, display
from gtts import gTTS

while True:
    text = input("입력: ").strip()

    if text == "종료":
        print("종료합니다.")
        break

    if text:
        tts = gTTS(text=text, lang="ko")
        tts.save("temp.mp3")
        display(Audio("temp.mp3", autoplay=True))
```
