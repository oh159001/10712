```python
a = int(input())
b = int(input())
if a<b:
  while a<b+1:
    print(a,end=' ')
    a+=1
elif a>b:
  while b<a+1:
    print(b,end=' ')
    b+=1
```

```python
a,b = map(int,input().split())
c = max(a,b)
d = min(a,b)
while d<=c:
  if d%2==0:
    print(d,end=' ')
  d+=1
```
