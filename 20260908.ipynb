```python
r = int(input('빨간색접시의 개수:'))
y = int(input('노란색접시의 개수:'))
b = int(input())
g = int(input())
print('====회전 초밥 영수증====')
print('  접시 단가 개수 합계')
print(' 빨간색 1000', r,r*1000 )
print(' 노란색 2000', y,y*2000 )
print(' 파란색 3000', b,b*3000 )
print(' 초록색 4000', g,g*4000 )
print(' 총가격       ',(r*1000)+(y*2000)+(b*3000)+(g*4000))
print('====================')
```
